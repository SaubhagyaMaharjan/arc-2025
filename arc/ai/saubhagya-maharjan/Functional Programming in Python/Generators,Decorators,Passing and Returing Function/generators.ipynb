```python
# generators produces sequence of values when iterated over.
#useful when we produce large no. of sequence but does not want to store all of them in memory at once.
# generator function is define like normal func but uses yiels statement 
```

```python
def my_generator(n):
    for i in range(n):
        yield i
# When we call the generator function, it returns an iterator object
output = my_generator(5) 
print(output)
```

```python
# Using a for loop
for item in output:
    print(item)
```

```python
# IMP: Generator will generate items one by one and only once
# So if we try to loop over the generator again, it will no longer generate items

for item in output:
    print(f"Again:{item}")
```

```python
# Creating Generators Using Generator Expression
# It uses parenthesis () instead of square bracket.
# Syntax:(expression for item in iterable)
  
my_iterator = (x  for x in range(5))

# Using the generator expression
for value in my_iterator:
    print(value)
```
