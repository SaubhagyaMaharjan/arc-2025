```python
# reduce applies cumulatively(left to right) to the items of an iterable.
# useful for performing cumulative operations on a collection of data.
#Syntax: reduce (function, iterable)
```

```python
from functools import reduce
def add(x,y):
    return x + y
numbers = [1,2,3,4,5]
sum = reduce(add,numbers)
print(sum)
```

```python
# reduce function with lambda function
from functools import reduce
numbers = [0,1,2,3]
sum = reduce (lambda x, y: x + y,numbers)
print(sum)
```
