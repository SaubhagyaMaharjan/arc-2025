```python
# map allow efficient and concise data processing
# Syntax : map(function , *iterable)
def square(x):
    return x * x

number = [2,3,4,5]
sqr_num = map(square, number)
sqr_num = list(sqr_num)
print(sqr_num)
```

```python
# map function with lambda function

number = [1,2,3,4]
sqr_num = map(lambda x:x*x ,number)
sqr_num = list(sqr_num)
print(sqr_num)
```
