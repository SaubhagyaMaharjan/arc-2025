```python
digits = [1,2,3]
```

```python
def plus_one(digits):
    n = len(digits)

    for i in range (n-1,-1,-1):
        if digits[i]<9:
            digits[i] +=1
            return digits
        else:
            digits[i]=0
    return[i] + digits
```

```python
final = plus_one(digits)
```
