```python
# Unpacking a tuple
point = (3,4)
x , y = point
print(y)
print(x)
```

```python
# Unpacking a list
colors = ["red", "green", "blue"]
r, g, b = colors
print(r) 
print(g) 
print(b) 
```

```python
# Extended unpacking
numbers = [1, 2, 3, 4, 5]
first, second, *rest = numbers
print(first)  
print(second)  
print(rest)   
```

```python
# Using * in the middle
numbers = [1, 2, 3, 4, 5]
first, *middle, last = numbers
print(first)  
print(middle)  
print(last)  
```
