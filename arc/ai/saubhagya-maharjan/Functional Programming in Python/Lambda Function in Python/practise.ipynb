```python
# normal function
def add(x,y):
    return x +y
sum = add(10 ,7)
print(sum)
```

```python
#lambda function; anonymous function because it has no name function, uses lambda keyword
# Syntax : 
# lambda arguments: expression

add = lambda x , y : x+y
print(add(5,4))
```

```python
check_odd_even = lambda x:"Even" if x % 2 == 0 else "Odd"
print(check_odd_even(17))
print(check_odd_even(10))
```
