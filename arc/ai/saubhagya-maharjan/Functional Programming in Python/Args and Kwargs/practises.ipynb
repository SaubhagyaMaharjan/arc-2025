```python
# *args and **kwargs are used when number of arguments are not fixed.
# * means an argument can be passed to function as a tuple.
# ** means an argument can be passed to a function as a dictionary.
# *args = positional argument

def delivery(order, *location):
    print(f"I have {order} 5 items.")
    print(f"Deliver at:")
    for point in location:
        print(f"-{point}")
```

```python
delivery('pants','chandragiri','kisipide')
```

```python
# **kwargs = keyword aeguments passed with keys and values

def bought(foods , **details):
    print(f"I buy {foods}.")
    for key,values in details.items():
        print(f"{key} = {values}")
```

```python
bought('mango', order_item = True , tips = '10rs')
```

```python
def travel(costs, *destination , **exp):
    print(f"It took Rs.{costs} for destination:")
    for list in destination:
        print(f"-{list}")
    print("Experiences:")
    for key , values in exp.items():
        print(f"{key} = {values}")
    
```

```python
travel( 20000,5000,'bandipur','pokhara','mustang', places = "Babal" , time_taken = '7 days' , fun = 'unlimited')
```
