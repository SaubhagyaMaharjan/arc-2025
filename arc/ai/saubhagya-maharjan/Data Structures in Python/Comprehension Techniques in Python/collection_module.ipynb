```python
# First import the deque class from collections
from collections import deque
```

```python
# Create an instance of deque
d = deque([1, 2, 3])
```

```python
#operations in deque
d.append(4)
d.appendleft(0)
```

```python
print(d)
```

```python
d.pop()
d.popleft()
```

```python
print(d)
```

```python
d.insert(1,2.5)
```
