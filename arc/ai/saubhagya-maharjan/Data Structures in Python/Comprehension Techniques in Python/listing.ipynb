```python
list_num = [1,2,3,4,5]
list_sqr_of_num =[]
```

```python
for num in list_num:
    sqr_num = num ** 2
    list_sqr_of_num.append(sqr_num)
```

```python
print(list_sqr_of_num)
```

```python
list_square_of_num = [num **2 for num in list_num]
```

```python
print(list_square_of_num)
```

```python
list_sqr_of_even_num = [num** 2 for num in list_num if num% 2 ==0]
```

```python
print(list_sqr_of_even_num)
```
