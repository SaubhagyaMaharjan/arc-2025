```python
original_dict = {'a': 1, 'b': 2, 'c': 3}
swapped_dict = {}
```

```python
for key, value in original_dict.items():
    swapped_dict[value] = key
```

```python
print(swapped_dict)
```

```python
swapped_dictt ={value:key for key, value in original_dict.items()}
```

```python
print(swapped_dictt)
```

```python
# Dictionary comprehension to calculate square of list of numbers
numbers = [1, 2, 3, 4, 5]
```

```python
sqr_dict = {x:x**2 for x in numbers}
```

```python
print(sqr_dict)
```
