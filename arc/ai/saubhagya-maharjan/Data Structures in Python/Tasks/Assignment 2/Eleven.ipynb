```python
import pandas as pd
```

```python


grades_dict = {
        "Alice": {"Math": 90, "Science": 85, "Literature": 88},
        "Bob": {"Math": 78, "Science": 82, "Literature": 80},
        "Charlie": {"Math": 92, "Science": 91, "Literature": 94}
}
```

```python
percentages = {}
```

```python
for student, subjects in grades_dict.items():
    total = sum(subjects.values())
    count = len(subjects)
    percentage = total / count
    percentages[student] = round(percentage, 2)
```

```python
for student, percent in percentages.items():
    print(f"{student}: {percent}")
```
