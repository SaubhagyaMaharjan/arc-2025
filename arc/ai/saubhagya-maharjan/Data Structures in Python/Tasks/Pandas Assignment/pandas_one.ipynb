```python
import pandas as pd
```

```python

data ={
"name":[],
"roll":[],
"address":[],
"age":[]
}
```

```python
for i in range(10):
    print(f"Enter details of student:{i + 1}")
    name = input("Name:")
    roll = input("Roll:")
    address = input("Address:")
    age = int(input("Age:"))

    data['name'].append(name)
    data['roll'].append(roll)
    data['address'].append(address)
    data['age'].append(age)
```

```python
dataframe = pd.DataFrame(data)
```

```python
def category(age):
    if age <= 15:
        return 'Junior'
    elif age > 15 and age < 25:
        return 'Mid'
    else:
        return 'Senior'

dataframe["category"] = dataframe['age'].apply(category)
```

```python

avg_age = dataframe['age'].mean()
most_freq = dataframe['category'].mode()[0]
```

```python
print(dataframe)
print(f"Average age: {avg_age}")
print(f"Most frequent Category is {most_freq}")
```
